```python
import datetime

print('<디데이(D-day) 프로그램>')
print()

year = int(input('연도를 입력하세요(예: 2026): '))
month = int(input('월을 입력하세요(1-12): '))
day = int(input('일을 입력하세요(1-31): '))

date = datetime.date(year, month, day)

today = datetime.date.today()

dday = (date - today).days

print(f'{year}년 {month}월 {day}일은 {dday}일 남았습니다.')
```

```python
red_price = 1000
yellow_price = 2000
blue_price = 3000
green_price = 4000

red_count = int(input('빨간색 접시의 개수를 입력하세요: '))
yellow_count = int(input('노란색 접시의 개수를 입력하세요: '))
blue_count = int(input('파란색 접시의 개수를 입력하세요: '))
green_count = int(input('초록색 접시의 개수를 입력하세요: '))

red_total = red_price * red_count
yellow_total = yellow_price * yellow_count
blue_total = blue_price * blue_count
green_total = green_price * green_count

total_price = red_total + yellow_total + blue_total + green_total

print()
print('===== 회전 초밥 영수증 =====')
print(f'{"접시":<8}{"단가":<8}{"개수":<8}{"합계":<8}')
print(f'{"빨간색":<8}{red_price:<8,}{red_count:<8}{red_total:<8,}')
print(f'{"노란색":<8}{yellow_price:<8,}{yellow_count:<8}{yellow_total:<8,}')
print(f'{"파란색":<8}{blue_price:<8,}{blue_count:<8}{blue_total:<8,}')
print(f'{"초록색":<8}{green_price:<8,}{green_count:<8}{green_total:<8,}')
print(f'총 가격: {total_price:,}원')
print('============================')
```

```python
print('<적금 계산기>')

deposit = int(input('월 납입액을 입력하세요 (원): '))
rate = float(input('연 이자율을 입력하세요 (%): '))
period = int(input('적금 기간을 입력하세요 (월): '))

total = deposit * period

interest = (deposit * rate / 100) * (period / 12)

final = total + interest

# 정수 부분만 출력
total = int(total)
interest = int(interest)
final = int(final)

print()
print(f'월 납입액 합계: {total:,}원')
print(f'적금 기간 발생한 총 이자금액: {interest:,}원')
print(f'적금 만기 시 받을 총금액: {final:,}원')
```
