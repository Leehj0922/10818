```python
def X (a,b,c):
  if b=='+':
    return a+c
  elif b=='-':
    return a-c
  elif b=='*':
    return a*c
  elif b=='/':
    if c==0:
      return '오류'
    return a/c
  else :
    return '오류'

print(X(3,'+',5))
print(X(3,'-',5))
print(X(3,'*',5))
print(X(3,'/',5))
```
