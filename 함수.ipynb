```python
def odd_even(num):
  if num %2==0:
    return '짝수'
  else:
    return '홀수'


int_num=int(input('정수를 입력'))
print(odd_even(int_num\))
```

```python
def tri_area(b,h):
  area=0.5 * b * h
  return area

b=int(input('밑변입력'))
h=int(input('높이입력'))
result=tri_area(b,h)
print(f'삼각형의 넓이는{result}입니다')
```

```python
concat=lambda s1,s2: s1+''+s2
result=concat('파이썬','프로그래밍')
print('문자열 합치기:',result)
```
