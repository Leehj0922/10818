```python
from gtts  import gTTS
from IPython.display import Audio
while True:
  text=input('텍스트를 입력:')
  if text =='종료':
    print('TTS 종료')
    break
  tts=gTTS(text,lang='en')
  tts.save('output.mp3')
  display(Audio('output.mp3',autoplay=True))
```
